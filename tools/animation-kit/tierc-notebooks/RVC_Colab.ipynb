## Voice-model workflow (the leprechaun pipeline)
1. Get 10+ min of clean approved voice (the V1 leprechaun line looped/extended works as a seed, but more variety is better).
2. Train tab → name it (e.g. `leprechaun_v1`) → train (~30-60 min on T4).
3. Infer tab → convert any new dialogue line into the leprechaun voice.
4. Download the `.pth` → commit to the repo so the voice is never lost (salvage law).
